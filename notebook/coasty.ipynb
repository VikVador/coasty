**Figure 2** — Spatial distribution of hypoxic profiles (1950–present + per decade) with Diaz & Rosenberg (2008) historical hypoxic zones superimposed.
Note that: 
1. Diaz & Rosenberg put their hypoxia threshold at 2 ml/L, whereas ours is 63 micromol/kg;
2. Currently the plot represent the location of ALL the element of the hypoxic dataset, included those classified as improved or eutrophic and not as hypoxic;
3. We associated the hypoxia_current label to our color classification. This is of course partially arbitrary, since in one case we are considering studies in an area and in the other frequency of collected observations, but it can help us to have a better idea of the comparison between the two datasets; connecting to point 2, some of the profiles where the hypoxia_current is empty are not classified as hypoxic, but as eutrophic. We included them as red triangles, to evaluate if we want to remove.
